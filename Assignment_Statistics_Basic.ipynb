#### Question 1: What is statistics, and why is it important?

#### Answer 1:
 Statistics is the science of collecting, organizing, analyzing, interpreting, and presenting data.

 Importance: It helps in decision-making, understanding trends, making predictions, and validating scientific theories across fields like business, medicine, social sciences, and engineering.

---------------------------------------------------

#### Question 2: What are the two main types of statistics?

#### Answer 2:
 Descriptive Statistics: Summarizes data (e.g., mean, median, charts).

 Inferential Statistics: Makes predictions or generalizations about a population based on a sample.

------------------------------------------------------

#### Question 3: What are descriptive statistics?

#### Answer 3:
 Descriptive statistics describe and summarize data using:

 Measures of central tendency (mean, median, mode)
 Measures of dispersion (range, variance, standard deviation)
 Tables, charts, and graphs 

------------------------------------------------------

#### Question 4: What is inferential statistics?

#### Answer 4:  
 It involves making inferences about populations using data drawn from a sample. Tools include:

 Hypothesis testing
 Confidence intervals
 Regression analysis

-----------------------------------------------------

#### Question 5: What is sampling in statistics?

#### Answer 5: 
Sampling is the process of selecting a subset (sample) from a larger group (population) to estimate characteristics of the whole population.

---------------------------------------------------

#### Question 6:  What are the different types of sampling methods?

#### Answer 6:
 Random sampling: Each member has an equal chance; reduces bias.
 Non-random sampling: Selection is based on convenience or judgment; may introduce bias.

 Numerical data: Interval, Ratio

----------------------------------------------------

#### Question 10: xplain nominal, ordinal, interval, and ratio levels of measurement.

#### Answer 10:
 Nominal: Categories (e.g., gender, colors)

 Ordinal: Ordered categories (e.g., rankings)

 Interval: Numeric with equal intervals, no true zero (e.g., temperature in °C)

 Ratio: Numeric with a true zero (e.g., weight, height)

#### Question 15: What is standard deviation, and why is it important?

#### Answer 15:
 Standard deviation is the square root of variance. It indicates how spread out the values are in a dataset. It’s easier to interpret as it's in the same units as the data.

----------------------------------------------------------

#### Question 16: Define and explain the term range in statistics.

#### Answer 16:
 Range = Max value - Min value. It gives a quick sense of data spread.

----------------------------------------------------------

#### Question 17: What is the difference between variance and standard deviation?

#### Answer 17:
 Variance: Average squared deviation from the mean.

 Standard Deviation: Square root of variance; more interpretable.

-----------------------------------------------------

#### Question 18: What is skewness in a dataset?


#### Answer 18:
 Skewness measures the asymmetry of the data distribution.

-----------------------------------------------

#### Question 19:  What does it mean if a dataset is positively or negatively skewed?

#### Answer 19: 
Positive skew: Tail on the right (mean > median)

Negative skew: Tail on the left (mean < median)

---------------------------------------------------

#### Question 20: Define and explain kurtosis.

#### Answer 20:
 Kurtosis measures the "tailedness" of the distribution.

High kurtosis: Heavy tails, more outliers.

Low kurtosis: Light tails, fewer outliers.

--------------------------------------------------------


#### Question 21: What is the purpose of covariance?

#### Answer 21:
 Covariance indicates the direction of the relationship between two variables (positive or negative).

-------------------------------------------------

#### Question 22: What does correlation measure in statistics?

#### Answer 22:
 Correlation measures both the strength and direction of a linear relationship between two variables (values range from -1 to +1).

---------------------------------------------------------

#### Question 23: What is the difference between covariance and correlation?

#### Answer 23: 
Covariance: Direction of the relationship.

Correlation: Direction and standardized strength (scale-free).
